```python
%sql
-- ============================================================================
-- Pipeline Step: Gold Dimension - dim_editor_type
-- Description: Conformed dimension defining Bot vs Human editor classifications.
-- Dependencies: Autonomous.
-- ============================================================================

USE CATALOG dbr_dev;
USE SCHEMA wikimediademo_gold;

CREATE OR REPLACE TABLE dbr_dev.wikimediademo_gold.dim_editor_type AS
SELECT 
    1 AS editor_type_key, 
    true AS is_bot, 
    'Bot' AS editor_type_desc
UNION ALL
SELECT 
    2 AS editor_type_key, 
    false AS is_bot, 
    'Human' AS editor_type_desc;

```
